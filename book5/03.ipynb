```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
num_toss = 10000000
x = np.random.randint(1,7,size = num_toss)
y = np.random.randint(1,7,size = num_toss)
```

```python
num_toss_array = np.arange(1,num_toss + 1)
```

```python
sum_6 = np.cumsum((x+y) == 6) #计算和为6的累计次数 计算里面生成的布尔数组的累加和注意返回的形状不变
```

```python
prob_sum_6 = sum_6 / num_toss_array
```

```python
fig,ax = plt.subplots()
plt.plot(num_toss_array,prob_sum_6)

ax.set_xscale('log')
plt.xlabel('Number of tosses')
plt.ylabel('Probability')

ax.grid(linestyle = '--',linewidth = .25,color = '0.5')
```

```python
num_toss = 10000
toss_times = 10 #每组10个
```

```python
toss_results = np.random.randint(0,2,size = (num_toss,toss_times))
```

```python
sum_results = np.sum(toss_results,axis = 1)
```

```python
results_exact_6 = (sum_results == 6)
```

```python
results_at_least_6 = (sum_results >= 6)
```

```python
num_toss_array = np.arange(1,num_toss + 1)
```

```python
count_6 = np.cumsum(results_exact_6)
count_6_at_least = np.cumsum(results_at_least_6)
```

```python
prob_count_6 = count_6 / num_toss_array
prob_count_6_at_least = count_6_at_least / num_toss_array
```

```python
fig,ax = plt.subplots()
plt.plot(num_toss_array,prob_count_6)

ax.set_xscale('log') # 设置x轴为对数坐标
plt.xlabel('Number of tosses'); # 设置x轴标签
plt.ylabel('Probability') # 设置y轴标签
ax.grid(linestyle='--', linewidth=0.25, color=[0.5,0.5,0.5]) # 设置网格线样式
```

```python
fig,ax = plt.subplots()
plt.plot(num_toss_array,prob_count_6_at_least)

ax.set_xscale('log') # 设置x轴为对数坐标
plt.xlabel('Number of tosses'); # 设置x轴标签
plt.ylabel('Probability') # 设置y轴标签
ax.grid(linestyle='--', linewidth=0.25, color=[0.5,0.5,0.5]) # 设置网格线样式
```
